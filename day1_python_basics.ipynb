## 4. If / Else - Cleaning logic

```python
sales = 250.50
profit = -20.0

if profit > 0:
    print("Profitable order")
elif profit == 0:
    print("Break-even")
else:
    print("Loss - needs attention!")

# TODO: Write logic: if sales > 500 -> "High Value", else "Low Value"
if sales > 500:
    print("High value")
else: 
    print("Low Value")
```

## 5. Functions - Reuse your cleaning code

```python
def clean_price(price_str):
    """Turns '$250.50' into 250.5"""
    cleaned = price_str.replace("$", "").replace(",", "")
    return float(cleaned)

print(clean_price("$250.50"))
print(clean_price("$1,200.00"))

# TODO: Write a function profit_margin(sales, profit) that returns margin %
def profit_margin(sales, profit):
    return (profit / sales * 100) if sales != 0 else 0

print(profit_margin(250.5, 40.2))
```
